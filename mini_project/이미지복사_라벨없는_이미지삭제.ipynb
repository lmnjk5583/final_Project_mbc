```python
import os
import shutil
from pathlib import Path

# ============================================================
# 1. 경로 설정
# ============================================================

# 원본 경로
SRC_IMAGE_DIR = r"N:\개인\박대원\0211~0313_miniproject\highway-anomaly-detection\data\normal\images"
SRC_LABEL_DIR = r"N:\개인\박대원\0211~0313_miniproject\highway-anomaly-detection\data\normal\labels"

# 이동할 경로
DST_IMAGE_DIR = r"N:\개인\대원&수빈\최종 프로젝트\capture\images"
DST_LABEL_DIR = r"N:\개인\대원&수빈\최종 프로젝트\capture\labels"

# 확장자 설정
IMAGE_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff'}
LABEL_EXT = '.txt'  # YOLO 형식 기준 (.xml로 변경 가능)

# ============================================================
# 2. 이동할 폴더 생성
# ============================================================
os.makedirs(DST_IMAGE_DIR, exist_ok=True)
os.makedirs(DST_LABEL_DIR, exist_ok=True)

# ============================================================
# 3. 이미지 파일 이동 (복사)
# ============================================================
img_count = 0
for f in os.listdir(SRC_IMAGE_DIR):
    if Path(f).suffix.lower() in IMAGE_EXTS:
        src = os.path.join(SRC_IMAGE_DIR, f)
        dst = os.path.join(DST_IMAGE_DIR, f)
        shutil.copy2(src, dst)  # copy2: 메타데이터 보존 / move로 변경 가능
        img_count += 1

print(f"✅ 이미지 {img_count}개 복사 완료 → {DST_IMAGE_DIR}")

# ============================================================
# 4. 라벨 파일 이동 (복사)
# ============================================================
label_count = 0
for f in os.listdir(SRC_LABEL_DIR):
    if Path(f).suffix.lower() == LABEL_EXT:
        src = os.path.join(SRC_LABEL_DIR, f)
        dst = os.path.join(DST_LABEL_DIR, f)
        shutil.copy2(src, dst)
        label_count += 1

print(f"✅ 라벨 {label_count}개 복사 완료 → {DST_LABEL_DIR}")

# ============================================================
# 5. 라벨 없는 이미지 삭제
# ============================================================
# 라벨 파일의 이름(확장자 제외) 목록 수집
label_names = set()
for f in os.listdir(DST_LABEL_DIR):
    if Path(f).suffix.lower() == LABEL_EXT:
        label_names.add(Path(f).stem)

print(f"\n📁 이동된 이미지: {img_count}개")
print(f"📁 이동된 라벨:  {label_count}개")
print(f"🔍 라벨 없는 이미지 탐색 중...\n")

deleted_count = 0
kept_count = 0

for f in os.listdir(DST_IMAGE_DIR):
    file_path = os.path.join(DST_IMAGE_DIR, f)
    
    if Path(f).suffix.lower() in IMAGE_EXTS:
        stem = Path(f).stem  # 파일명 (확장자 제외)
        
        if stem not in label_names:
            os.remove(file_path)
            print(f"  🗑️  삭제: {f}")
            deleted_count += 1
        else:
            kept_count += 1

# ============================================================
# 6. 결과 출력
# ============================================================
print(f"\n{'='*50}")
print(f"📊 처리 결과")
print(f"{'='*50}")
print(f"  유지된 이미지: {kept_count}개")
print(f"  삭제된 이미지: {deleted_count}개")
print(f"  최종 이미지 수: {kept_count}개")
print(f"  라벨 파일 수:  {len(label_names)}개")
print(f"{'='*50}")
```
