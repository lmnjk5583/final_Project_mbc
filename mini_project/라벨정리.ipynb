```python
import os
from pathlib import Path

LABEL_DIRS = [
    Path(r"N:\개인\대원&수빈\최종 프로젝트\capture\labels"),
    # Path(r"N:\개인\대원&수빈\최종 프로젝트\capture\train\labels"),
    # Path(r"N:\개인\대원&수빈\최종 프로젝트\capture\val\labels"),
    # Path(r"N:\개인\대원&수빈\최종 프로젝트\capture\test\labels"),
]

total_fixed = 0
for label_dir in LABEL_DIRS:
    if not label_dir.exists():
        continue
    fixed = 0
    for txt_file in label_dir.glob("*.txt"):
        lines = txt_file.read_text(encoding='utf-8').strip().splitlines()
        new_lines = []
        modified = False
        for line in lines:
            parts = line.strip().split()
            if len(parts) == 6:
                new_lines.append(' '.join(parts[:5]))
                modified = True
            else:
                new_lines.append(line.strip())
        if modified:
            txt_file.write_text('\n'.join(new_lines) + '\n', encoding='utf-8')
            fixed += 1
    print(f"📁 {label_dir.name}: {fixed}개 수정")
    total_fixed += fixed

print(f"\n✅ 총 {total_fixed}개 파일 수정 완료")
```
